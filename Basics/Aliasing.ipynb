In Python, aliasing means that two different variable names refer to the same object in memory. When you change the object using one variable, the change is reflected when you access it through the other variable, because they are just "aliases" for the same underlying data.

```python
x = [10, 20, 40]
y = x
print(x is y)
```

```python
x = [10, 20, 40]
y = x
y.append(60)
print(x)
print(y)
```
