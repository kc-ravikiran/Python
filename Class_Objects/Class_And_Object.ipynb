```python
class Vehicle:
    category = "Transport"

    def __init__(self, name, color, price):
        self.name = name        # Instance attribute
        self.color = color      # Instance attribute
        self.price = price      # Instance attribute

    # Method to display vehicle details
    def display_info(self):
        return f"{self.color} {self.name} worth ₹{self.price}"

# Create objects (instances of Vehicle class)
car1 = Vehicle("Sierra", "Red", 1500000)
car2 = Vehicle("Tata Punch", "Blue", 1400000)

# Access object methods
print(car1.display_info())   
print(car2.display_info()) 
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return f"Deposited ₹{amount}. New balance: ₹{self.balance}"

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds!"
        self.balance -= amount
        return f"Withdrew ₹{amount}. New balance: ₹{self.balance}"

# Create objects
account1 = BankAccount("Kiran", 5000)
account2 = BankAccount("Ravi", 10000)

print(account1.deposit(2000))  
print(account2.withdraw(3000))  
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def annual_bonus(self):
        return self.salary * 0.10

# Objects
emp1 = Employee("Rahul", 50000)
emp2 = Employee("Sneha", 70000)

print(emp1.annual_bonus())  # 5000.0
print(emp2.annual_bonus())  # 7000.0
```
